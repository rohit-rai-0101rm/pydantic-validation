```python
from pydantic import BaseModel,Field,EmailStr,AnyUrl,Field,field_validator,model_validator,computed_field
from typing import List,Dict,Optional,Annotated

class PatientData(BaseModel):
    name:Annotated[str,Field(max_length=50,title="Nam of the patient",description='Give the name of the patient')]
    age:int=Field(gt=0,lt=100)
    email:EmailStr
    linkedin_url:AnyUrl
    height:float
    weight:Annotated[float,Field(gt=0,strict=True,description="Weight of the patient in kg")]
    married:Annotated[bool,Field(default=None,description="Is the patient married or not")]
    allergies:Annotated[Optional[List[str]],Field(max_length=5)]
    contact_info:dict[str,str]

    @computed_field
    @property
    def bmi(self) -> float:
        bmi=round(self.weight/(self.height**2),2)
        return bmi

    @model_validator(mode="after")
    def validate_emergency_contact(cls,model):
        if model.age>60 and 'emergency' not in model.contact_info:
            raise ValueError('Patients Older than 60 must have an emergency contact')
        return model

    @field_validator('email')
    @classmethod
    def email_validator(cls,value):
        valid_domains=['hdfc.com','icici.com']
        domain_name=value.split('@')[-1]
        if domain_name not in valid_domains:
            raise ValueError('Not a valid domain')
        return value
    @field_validator("name")
    @classmethod
    def transfrom_name(cls,value):
        return value.upper()
def add_patient_data(patient:PatientData):
   print(patient.email)
   print(patient.name)
   print(patient.age)
   print(patient.weight)

   print(patient.linkedin_url)
  
   print(patient.allergies)
   print(patient.contact_info)
   print(patient.married)
   print(patient.bmi)

def update_patient_data(patient:PatientData):

    print(patient.email)
    print(patient.name)
    print(patient.age)
    print(patient.weight)

    print(patient.linkedin_url)


    print(patient.allergies)
    print(patient.contact_info)
    print(patient.married)

    print("Data updated successfully in the databse")
patient_data={"name":"Bappy","email":"rohit@hdfc.com","allergies":["1","2","3","4","5"], "linkedin_url":"https://www.linkedin.com/feed/","age":75,"weight":80.4, "height":1.75,"married":"True","contact_info":{"email":"rohit@exalogic.co","phone":"+918144256428","emergency":"+918144256428"}}


patient_1=PatientData(**patient_data)
add_patient_data(patient_1)


```
