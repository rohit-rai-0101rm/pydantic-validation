```python
from pydantic import BaseModel,Field

class PatientData(BaseModel):
    name:str
    age:int
    weight:float
def add_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
def update_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
    print("Data updated successfully in the databse")
patient_data={"name":"Bappy","age":25,"weight":"80.4"}


patient_1=PatientData(**patient_data)
add_patient_data(patient_1)


```

```python
from pydantic import BaseModel,Field
from typing import List,Dict

class PatientData(BaseModel):
    name:str
    age:int
    weight:float
    married:bool=False
    allergies:List[str]
    contact_info:dict[str,str]
def add_patient_data(patient:PatientData):
    print(patient.name)
    print(patient.age)
    print(patient.weight)
def update_patient_data(patient:PatientData):

    print(patient)
    print(patient.name)
    print(patient.age)
    print(patient.weight)

    print(patient.allergies)
    print(patient.contact_info)
    print(patient.married)
    print("Data updated successfully in the databse")
patient_data={"name":"Bappy","age":25,"weight":"80.4","married":"True","allergies":["peanuts","dust"],"contact_info":{"email":"rohit@exalogic.co","phone":"+918144256428"}}


patient_1=PatientData(**patient_data)
add_patient_data(patient_1)


```
